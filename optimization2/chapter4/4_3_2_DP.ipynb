```python
# ナップザック問題をやってみる。
# アルゴリズム4.4について、図4.41の例をやってみる。

prices = [16, 19, 23, 28]
weights = [2, 3, 4, 5]
maxWeight = 7

# このリストの行は荷物のID。列はそのIDまでの袋の容量(パラメータ名は重さとしている)の合計。テーブルの値がその行のIDまでの荷物の価値の合計。
dpList = []

# step1 f(1,u)を計算する
tmpF = [[0, [0]] if u < weights[0] else [prices[0], [1]] for u in range(maxWeight + 1)]
dpList.append(tmpF)
print(dpList)

# step2 n-1回繰り返す。
# ここのオーダがn
for k in range(1, len(prices)):
    # step3 (4.63)式に従い、f(k,u)を求める。
    tmpF = []
    # ここのオーダがC
    for u in range(maxWeight + 1):
        # まずu < w_kの時
        if u < weights[k]:
            tmpF.append([dpList[k-1][u][0], dpList[k-1][u][1] + [0]])
        else:
            # 追加されないときの値
            notAddedValue = dpList[k-1][u][0]
            # 追加されるときの値。u-w_k >= 0は保証されている。
            addedValue = dpList[k-1][u - weights[k]][0] + prices[k]
            # maxをif文で考える。
            if notAddedValue > addedValue:
                tmpF.append([notAddedValue, dpList[k-1][u][1] + [0]])
            else:
                tmpF.append([addedValue, dpList[k-1][u - weights[k]][1] + [1]])
    dpList.append(tmpF)

for l in dpList:
    print(l)

print("最適値(袋の容量が44の時の最大の価値) : ", dpList[len(dpList) - 1][maxWeight][0])
print("最適値の時のx : ", dpList[len(dpList) - 1][maxWeight][1])
```

```python
# アルゴリズム4.5を実装する。
# アルゴリズム4.4でのオーダはO(nC)だったが、アルゴリズム4.5ではO(nP)になる。ここでWはwの分解能。
# 例えばwの分解能が0.1ならば、WはCの10倍になる。これが大きくなるのであれば以下のコードが有効に働く場合があるかもしれない。

import numpy as np

prices = [3, 4, 1, 2]
weights = [2, 3, 1, 3]
maxPrice = sum(prices)
maxWeight = 4

dpList = []

# step1 f(1,u)を計算する
tmpF = [[0, [0]]]
for v in range(1, maxPrice + 1):
    if v == prices[0]:
        tmpF.append([weights[0], [1]])
    else:
        tmpF.append([np.inf, [0]])
dpList.append(tmpF)
#print(dpList)

# step2 n-1回繰り返す。
for k in range(1, len(prices)):
    # step3 (4.63)式に従い、f(k,u)を求める。
    tmpF = []
    for v in range(maxPrice + 1):
        # まずv < p_kの時
        if v < prices[k]:
            tmpF.append([dpList[k-1][v][0], dpList[k-1][v][1] + [0]])
        else:
            # 追加されないときの値
            notAddedValue = dpList[k-1][v][0]
            # 追加されるときの値。v-p_k >= 0は保証されている。
            addedValue = dpList[k-1][v - prices[k]][0] + weights[k]
            # maxをif文で考える。
            if notAddedValue < addedValue:
                tmpF.append([notAddedValue, dpList[k-1][v][1] + [0]])
            else:
                tmpF.append([addedValue, dpList[k-1][v - prices[k]][1] + [1]])
    dpList.append(tmpF)

for i, l in enumerate(dpList):
    print("luggage ", i, " : weight ", weights[i], " : ", l)

# 制限を超えない最大のインデックスを探す。
i = 0
for weight, items in reversed(dpList[-1]):
    #print("weight : ", weight, " : items : ", items, " : i : ", i)
    if weight <= maxWeight:
         index = maxPrice - i
         break
    i += 1

print("最適値 : ", index)
print("最適値の時のx : ", dpList[-1][index][1])
```

```python
# アルゴリズム4.6について、図4.43の例をやってみる。
# 基本的にアルゴリズム4.4と同様でOK

# 汎用性が上がるように考慮
import sympy as sp
import operator
import numpy as np

class constraintObject:
    # 変数binで{0, 1}に制限されているか、自然数かを切り替える。
    # 個別に設定することも考えられるが、今はまとめて切り替える。
    # 個別に切り替えると実装上、変数との対応に工夫が必要。
    def __init__(self, expr, value, op, bin = False):
        # 制約式の左辺
        self.expr = expr
        # 制約式の右辺の値
        self.value = value
        # 比較方法。operator.eq, operator.leのみを受け付ける。
        # とはいうものの簡略化のため、異常処理は行わない。 
        self.operator = op
        self.bin = bin

        # 厳密には制約条件に含まれない変数があってもよいが、目的関数と制約条件に含まれる変数は等しいとする。
        self.variableList = list(expr.free_symbols)
        # 必ずしも順番に並んでいるわけではない。
        #print(self.variableList)

        self.coefs = []
        for v in self.variableList:
            # 係数を取得する。
            self.coefs.append(int(expr.coeff(v, 1)))
        #print(self.coefs)

# 変数は正の整数として、目的関数を最大化する動的計画法を実装する。
class dp:
    def __init__(self, expr, constraintObj):
        self.expr = expr
        self.constraintObj = constraintObj

        self.valiableList = constraintObj.variableList

        self.fs = []
        for v in self.valiableList:
            # 各変数を含む項を取り出す。
            terms_with_v = [term for term in expr.as_ordered_terms() if term.has(v)]

            # 再び式に戻す
            result = sum([exps for exps in terms_with_v])
            #print(result)
            self.fs.append(result)
        #print(self.fs)

        # この値が本のBの値に相当する。   
        self.listSize = constraintObj.value + 1

        self.makeDpList()

    def makeDpList(self):        
        self.dpList = []

        # step1 f(1,u)を計算する
        tmpF = [[-np.inf,[]]] * self.listSize
        
        if self.constraintObj.bin:
            rangeNum = 2
        else:
            rangeNum = self.listSize

        for i in range(rangeNum):
            if self.constraintObj.coefs[0] * i >= 0 and self.constraintObj.coefs[0] * i <= self.listSize:
                tmpF[self.constraintObj.coefs[0] * i] = [self.fs[0].evalf(subs = {self.valiableList[0] : i}), [{self.valiableList[0] : i}]]        
        
        self.dpList.append(tmpF)
        #print(self.dpList[0])

        # step2 n-1回繰り返す。
        # ここのオーダがn
        for k in range(1, len(self.valiableList)):
            # step3
            tmpF = [[-np.inf, []]] * self.listSize

            # ここのオーダがB
            for i in range(self.listSize):
                # ここもオーダがB
                for j in range(rangeNum):
                    index = int(i + self.constraintObj.coefs[k] * j)

                    if index >= 0 and index < self.listSize:
                        #print(k, i, j, index, tmpF[index][0], self.dpList[k - 1][i][0] + self.fs[k].evalf(subs = {self.valiableList[k] : j}), self.dpList[k - 1][i][0], self.fs[k].evalf(subs = {self.valiableList[k] : j}))
                        # この部分をすべての組み合わせで実施することで、maxを計算していることになる。
                        if tmpF[index][0] < self.dpList[k - 1][i][0] + self.fs[k].evalf(subs = {self.valiableList[k] : j}):
                            tmpF[index] = [self.dpList[k - 1][i][0] + self.fs[k].evalf(subs = {self.valiableList[k] : j}), self.dpList[k - 1][i][1] + [{self.valiableList[k] : j}]]

            self.dpList.append(tmpF)

    def getDpList(self):
        return self.dpList
    
    def getResult(self):
        if self.constraintObj.operator == operator.eq:
            return self.dpList[-1][-1]
        elif self.constraintObj.operator == operator.le:
            lastRow = self.dpList[-1]

            retValue = lastRow[0]
            for i in range(1, len(lastRow)):
                if retValue[0] < lastRow[i][0]:
                    retValue = lastRow[i]

            return retValue

# 変数の定義
x1, x2, x3, x4 = sp.symbols('x1 x2 x3 x4')
# 最大化する式の定義
# この式は各項が一つの変数からなる式とする。
# 3次式を含み、単調増加であるが、上に凸というわけでもない。ほかの項も上に凸出ないものを含む
objectiveFunction = 10 * sp.Abs(x1 - 1) + x2**2 + 2 * (x3 - 1)**2 + 60 / sp.Abs(2 * x4 - 9)
# 制約条件の左辺
# この式の係数は自然数とする。また、1次式とする。(そうしないとリストを作るのが難しそう。)
constraint = x1 + x2 + x3 + x4
constraintObj = constraintObject(constraint, 6, operator.eq)

dpObj = dp(objectiveFunction, constraintObj)

result = dpObj.getDpList()

for r in result:
    print(r)

result = dpObj.getResult()
print("最適値 : ", result[0], " : その時の変数 : ", result[1])

```

```python
# アルゴリズム4.4よりも計算量は多いが、同様に問題を解ける。

# 変数の定義
x1, x2, x3, x4 = sp.symbols('x1 x2 x3 x4')
# 最大化する式の定義
# この式は各項が一つの変数からなる式とする。
# 3次式を含み、単調増加であるが、上に凸というわけでもない。ほかの項も上に凸出ないものを含む
objectiveFunction = 16 * x1 + 19 * x2 + 23 * x3 + 28 * x4
# 制約条件の左辺
# この式の係数は自然数とする。また、1次式とする。(そうしないとリストを作るのが難しそう。)
constraint = 2 * x1 + 3 * x2 + 4 * x3 + 5 * x4
constraintObj = constraintObject(constraint, 7, operator.le, bin = True)

dpObj = dp(objectiveFunction, constraintObj)

result = dpObj.getDpList()

for r in result:
    print(r)

result = dpObj.getResult()
print("最適値 : ", result[0], " : その時の変数 : ", result[1])
```

```python
# 上記のデータに対して、アルゴリズム4.7を実装してみる。
import numpy as np

dpList = []
for i in range((end[0] - start[0]) + 1):
    dpList.append([None] * ((end[1] - start[1])+ 1))
#print(dpList)

# ステップ1
dpList[0][0] = [0, [(0, 0)]]  # 始点のコストは0

#print(graph)

# ここのオーダはO(m)
for i in range(1, (end[0] - start[0]) + 1):
    dpList[i][0] = [graph[(i - 1, 0)][(i, 0)] + dpList[i - 1][0][0], dpList[i - 1][0][1] + [(i, 0)]]
print(dpList)

# ステップ2
# ここのオーダはO(m)→全体でO(mn)
for row in range(1, (end[1] - start[1]) + 1):
    # (0, row)については(0, row - 1)からしか矢印が向かない。これが最小値になる。
    dpList[0][row] = [(dpList[0][row - 1][0] + graph[(0, row - 1)][(0, row)]), dpList[0][row - 1][1] + [(0, row)]]
    # ここのオーダはO(n)
    for col in range(1, (end[0] - start[0]) + 1):
        prevPos = [(col - 1, row - 1), (col, row - 1), (col - 1, row)]
        #print(row, col)
        currentValues = [dpList[c][r][0] + graph[(c, r)][(col, row)] for c, r in prevPos]

        minValue = np.inf
        # ステップ3
        # ここの計算量はO(1)
        for pos, val in zip(prevPos, currentValues):
            if minValue > val:
                minValue = val
                prev = pos

        dpList[col][row] = [(dpList[prev[0]][prev[1]][0] + graph[(prev[0], prev[1])][(col, row)]), dpList[prev[0]][prev[1]][1] + [(col, row)]]

print(dpList)
print("最短距離 : ", dpList[6][5][0], " : 最短経路 : ", dpList[6][5][1])
```
