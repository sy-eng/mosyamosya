```python
# ナップザック問題をやってみる。
# アルゴリズム4.4について、図4.41の例をやってみる。

prices = [16, 19, 23, 28]
weights = [2, 3, 4, 5]
maxWeight = 7

# このリストの行は荷物のID。列はそのIDまでの袋の容量(パラメータ名は重さとしている)の合計。テーブルの値がその行のIDまでの荷物の価値の合計。
dpList = []

# step1 f(1,u)を計算する
tmpF = [[0, [0]] if u < weights[0] else [prices[0], [1]] for u in range(maxWeight + 1)]
dpList.append(tmpF)
print(dpList)

# step2 n-1回繰り返す。
# ここのオーダがn
for k in range(1, len(prices)):
    # step3 (4.63)式に従い、f(k,u)を求める。
    tmpF = []
    # ここのオーダがC
    for u in range(maxWeight + 1):
        # まずu < w_kの時
        if u < weights[k]:
            tmpF.append([dpList[k-1][u][0], dpList[k-1][u][1] + [0]])
        else:
            # 追加されないときの値
            notAddedValue = dpList[k-1][u][0]
            # 追加されるときの値。u-w_k >= 0は保証されている。
            addedValue = dpList[k-1][u - weights[k]][0] + prices[k]
            # maxをif文で考える。
            if notAddedValue > addedValue:
                tmpF.append([notAddedValue, dpList[k-1][u][1] + [0]])
            else:
                tmpF.append([addedValue, dpList[k-1][u - weights[k]][1] + [1]])
    dpList.append(tmpF)

for l in dpList:
    print(l)

print("最適値(袋の容量が44の時の最大の価値) : ", dpList[len(dpList) - 1][maxWeight][0])
print("最適値の時のx : ", dpList[len(dpList) - 1][maxWeight][1])
```

```python
# アルゴリズム4.5を実装する。
# アルゴリズム4.4でのオーダはO(nC)だったが、アルゴリズム4.5ではO(nP)になる。ここでWはwの分解能。
# 例えばwの分解能が0.1ならば、WはCの10倍になる。これが大きくなるのであれば以下のコードが有効に働く場合があるかもしれない。

import numpy as np

prices = [3, 4, 1, 2]
weights = [2, 3, 1, 3]
maxPrice = sum(prices)
maxWeight = 4

dpList = []

# step1 f(1,u)を計算する
tmpF = [[0, [0]]]
for v in range(1, maxPrice + 1):
    if v == prices[0]:
        tmpF.append([weights[0], [1]])
    else:
        tmpF.append([np.inf, [0]])
dpList.append(tmpF)
#print(dpList)

# step2 n-1回繰り返す。
for k in range(1, len(prices)):
    # step3 (4.63)式に従い、f(k,u)を求める。
    tmpF = []
    for v in range(maxPrice + 1):
        # まずv < p_kの時
        if v < prices[k]:
            tmpF.append([dpList[k-1][v][0], dpList[k-1][v][1] + [0]])
        else:
            # 追加されないときの値
            notAddedValue = dpList[k-1][v][0]
            # 追加されるときの値。v-p_k >= 0は保証されている。
            addedValue = dpList[k-1][v - prices[k]][0] + weights[k]
            # maxをif文で考える。
            if notAddedValue < addedValue:
                tmpF.append([notAddedValue, dpList[k-1][v][1] + [0]])
            else:
                tmpF.append([addedValue, dpList[k-1][v - prices[k]][1] + [1]])
    dpList.append(tmpF)

for i, l in enumerate(dpList):
    print("luggage ", i, " : weight ", weights[i], " : ", l)

# 制限を超えない最大のインデックスを探す。
i = 0
for weight, items in reversed(dpList[-1]):
    #print("weight : ", weight, " : items : ", items, " : i : ", i)
    if weight <= maxWeight:
         index = maxPrice - i
         break
    i += 1

print("最適値 : ", index)
print("最適値の時のx : ", dpList[-1][index][1])
```
