```python
# ナップザック問題をやってみる。

prices = [16, 19, 23, 28]
weights = [2, 3, 4, 5]
maxWeight = 7

dpList = []

# step1 f(1,u)を計算する
tmpF = [[0, [0]] if u < weights[0] else [prices[0], [1]] for u in range(maxWeight + 1)]
dpList.append(tmpF)
print(dpList)

# step2 n-1回繰り返す。
for k in range(1, len(prices)):
    # step3 (4.63)式に従い、f(k,u)を求める。
    tmpF = []
    for u in range(maxWeight + 1):
        # まずu < w_kの時
        if u < weights[k]:
            tmpF.append([dpList[k-1][u][0], dpList[k-1][u][1] + [0]])
        else:
            # 追加されないときの値
            notAddedValue = dpList[k-1][u][0]
            # 追加されるときの値。u-w_k >= 0は保証されている。
            addedValue = dpList[k-1][u - weights[k]][0] + prices[k]
            # maxをif文で考える。
            if notAddedValue > addedValue:
                tmpF.append([notAddedValue, dpList[k-1][u][1] + [0]])
            else:
                tmpF.append([addedValue, dpList[k-1][u - weights[k]][1] + [1]])
    dpList.append(tmpF)

for l in dpList:
    print(l)

print("最適値 : ", dpList[len(dpList) - 1][maxWeight][0])
print("最適値の時のx : ", dpList[len(dpList) - 1][maxWeight][1])
```
